```python
import pandas as pd
import numpy as np

# 1. GENERATE SIMPLE PATIENT DATA
np.random.seed(42)
num_records = 100

print("--- STEP 1: GENERATING PATIENT RECOVERY DATA ---")

diagnoses = ['Cardiovascular', 'Neurological', 'Orthopedic', 'Respiratory', 'Oncology']

data = {
    'Patient_ID': np.arange(1001, 1001 + num_records),
    'Age': np.random.randint(18, 90, size=num_records),
    'Gender': np.random.choice(['Male', 'Female'], size=num_records),
    'Primary_Diagnosis': np.random.choice(diagnoses, size=num_records),
    'Hospital_Stay_Days': np.random.randint(2, 20, size=num_records),
    'Recovery_Score': np.random.randint(40, 100, size=num_records)
}

df = pd.DataFrame(data)

# Inject exactly 4 missing data points into Hospital_Stay_Days to practice cleaning
df.loc[[10, 25, 50, 75], 'Hospital_Stay_Days'] = np.nan

print("Data successfully created!")
print(df.head())


# 2. CLEAN MISSING DATA
print("\n--- STEP 2: CLEANING MISSING VALUES ---")
print(f"Missing values before cleaning: {df['Hospital_Stay_Days'].isna().sum()}")

# Fill missing stays with the overall average stay duration
average_stay = df['Hospital_Stay_Days'].mean()
df['Hospital_Stay_Days'] = df['Hospital_Stay_Days'].fillna(average_stay)

print(f"Missing values after cleaning: {df['Hospital_Stay_Days'].isna().sum()}")


# 3. CALCULATE BASIC NUMPY STATISTICS
print("\n--- STEP 3: CALCULATING BASIC STATISTICS ---")

# Convert columns to simple NumPy arrays
age_array = df['Age'].to_numpy()
score_array = df['Recovery_Score'].to_numpy()

print(f"Average Patient Age: {np.mean(age_array):.1f} years")
print(f"Minimum Recovery Score: {np.min(score_array)}")
print(f"Maximum Recovery Score: {np.max(score_array)}")


# 4. FILTER HIGH RISK PATIENTS (Logical Masking)
print("\n--- STEP 4: FILTERING HIGH-RISK PATIENTS ---")

# Filter: Patients older than 70 with a low recovery score (under 55)
high_risk_condition = (df['Age'] > 70) & (df['Recovery_Score'] < 55)
high_risk_patients = df[high_risk_condition]

print(f"Found {len(high_risk_patients)} high-risk senior patients:")
print(high_risk_patients[['Patient_ID', 'Age', 'Primary_Diagnosis', 'Recovery_Score']])


# 5. GROUP DATA BY DIAGNOSIS
print("\n--- STEP 5: DIAGNOSIS SUMMARY TABLE ---")

summary_table = df.groupby('Primary_Diagnosis').agg(
    Total_Patients=('Patient_ID', 'count'),
    Avg_Stay_Days=('Hospital_Stay_Days', 'mean'),
    Avg_Recovery_Score=('Recovery_Score', 'mean')
).round(1)

print(summary_table)


```
